# CALCULATE CHURN RATE BY DIFFERENT CUSTOMER SEGMENTS 

```python
# Define a function to calculate churn percentage
def calculate_churn_rate(data, column):

    # Calculate total customers in each category
    total_customers = data.groupby(column)["Churn"].count()

    # Calculate customers who churned in each category
    churned_customers = data[
        data["Churn"] == "Yes"
    ].groupby(column)["Churn"].count()

    # Calculate churn percentage
    churn_rate = (
        churned_customers / total_customers
    ) * 100

    # Create a summary table
    result = pd.DataFrame({
        "Total Customers": total_customers,
        "Churned Customers": churned_customers,
        "Churn Rate (%)": churn_rate
    })

    # Replace missing churn counts with zero
    result["Churned Customers"] = (
        result["Churned Customers"].fillna(0)
    )

    # Sort from highest to lowest churn rate
    result = result.sort_values(
        by="Churn Rate (%)",
        ascending=False
    )

    return result
```

# APPLY THE FUNCTION 

```python
# Calculate churn rate by contract
print("Churn by Contract:")
display(calculate_churn_rate(df, "Contract"))

# Calculate churn rate by internet service
print("Churn by Internet Service:")
display(calculate_churn_rate(df, "InternetService"))

# Calculate churn rate by payment method
print("Churn by Payment Method:")
display(calculate_churn_rate(df, "PaymentMethod"))

# Calculate churn rate by gender
print("Churn by Gender:")
display(calculate_churn_rate(df, "gender"))

# Calculate churn rate by tech support
print("Churn by Tech Support:")
display(calculate_churn_rate(df, "TechSupport"))
```

# IDENTIFY HIGH RISK CUSTOMER SEGMENTS 

```python
# Create a list of important customer categories
important_features = [
    "Contract",
    "InternetService",
    "PaymentMethod",
    "TechSupport",
    "OnlineSecurity",
    "Partner",
    "Dependents"
]

# Loop through each feature
for feature in important_features:

    # Calculate churn rate for the feature
    result = calculate_churn_rate(df, feature)

    # Identify the category with the highest churn rate
    highest_churn_category = result.index[0]

    # Display the result
    print("\nFeature:", feature)

    print(
        "Highest churn category:",
        highest_churn_category
    )

    print(
        "Churn rate:",
        round(
            result.iloc[0]["Churn Rate (%)"],
            2
        ),
        "%"
    )
```

# CREATE OUTLIER DETECTION FUNCTION 

```python
# Define a function to detect outliers using IQR
def detect_outliers(data, column):

    # Calculate the first quartile
    Q1 = data[column].quantile(0.25)

    # Calculate the third quartile
    Q3 = data[column].quantile(0.75)

    # Calculate the interquartile range
    IQR = Q3 - Q1

    # Calculate the lower boundary
    lower_limit = Q1 - 1.5 * IQR

    # Calculate the upper boundary
    upper_limit = Q3 + 1.5 * IQR

    # Identify observations outside the boundaries
    outliers = data[
        (data[column] < lower_limit) |
        (data[column] > upper_limit)
    ]

    # Return the outlier information
    return {
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outlier Count": len(outliers),
        "Outlier Percentage": (
            len(outliers) / len(data)
        ) * 100
    }
```

# OUTLIER IN NUMERICAL COLUMN 

```python
# Select numerical columns for outlier detection
outlier_columns = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

# Create an empty list to store results
outlier_results = []

# Loop through each numerical column
for col in outlier_columns:

    # Detect outliers
    result = detect_outliers(df_clean, col)

    # Store the result
    outlier_results.append(result)

# Convert results into a DataFrame
outlier_report = pd.DataFrame(outlier_results)

# Display the outlier report
display(outlier_report)
```

```python
# Define the internet-related add-on columns
internet_service_columns = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

# Count how many additional internet services each customer uses
df_fe["InternetAddOnCount"] = (
    df_fe[internet_service_columns] == "Yes"
).sum(axis=1)

# Display the original columns and the new feature
display(
    df_fe[
        internet_service_columns +
        ["InternetAddOnCount"]
    ].head(10)
)
```

```python
# Create the average charges per tenure month feature
df_fe["AverageChargesPerTenureMonth"] = np.where(

    # Check whether tenure is greater than zero
    df_fe["tenure"] > 0,

    # Divide total charges by tenure
    df_fe["TotalCharges"] / df_fe["tenure"],

    # Assign zero for zero-tenure customers
    0
)

# Display the original and new features
display(
    df_fe[
        [
            "tenure",
            "MonthlyCharges",
            "TotalCharges",
            "AverageChargesPerTenureMonth"
        ]
    ].head(15)
)
```

# CREATE CUSTOMER SEGMENT 

```python
# Define a function to classify customers
def customer_segment(row):

    # Identify whether the customer is new
    is_new = row["tenure"] <= 12

    # Identify whether the monthly charge is high
    is_high_charge = row["MonthlyCharges"] > 70

    # New customer with high monthly charges
    if is_new and is_high_charge:
        return "New - High Billing"

    # New customer with lower monthly charges
    elif is_new and not is_high_charge:
        return "New - Low Billing"

    # Established customer with high monthly charges
    elif not is_new and is_high_charge:
        return "Established - High Billing"

    # Established customer with lower monthly charges
    else:
        return "Established - Low Billing"


# Apply the function to each customer
df_fe["CustomerSegment"] = df_fe.apply(
    customer_segment,
    axis=1
)

# Display the customer segments
display(
    df_fe[
        [
            "tenure",
            "MonthlyCharges",
            "CustomerSegment"
        ]
    ].head(20)
)
```

# IMPORT THE REQUIRED LIBRARIES 

```python
# Import pandas for working with tables and datasets
import pandas as pd

# Import numpy for numerical operations
import numpy as np

# Import train_test_split to divide data into training and testing sets
from sklearn.model_selection import train_test_split

# Import SimpleImputer to handle missing values
from sklearn.impute import SimpleImputer

# Import StandardScaler to scale numerical features
from sklearn.preprocessing import StandardScaler

# Import OneHotEncoder to convert categorical features into numbers
from sklearn.preprocessing import OneHotEncoder

# Import Pipeline to combine preprocessing steps
from sklearn.pipeline import Pipeline

# Import ColumnTransformer to apply different preprocessing to different columns
from sklearn.compose import ColumnTransformer

# Import joblib to save reusable Python objects
import joblib

# Import warnings to manage warning messages
import warnings

warnings.filterwarnings("ignore")
```

# LOAD THE ENGINEERED DATASET 

```python
# Load the engineered dataset
df = pd.read_csv(
    "data/processed/engineered_telco_customer_churn.csv"
)

# Display the first five rows
df.head()
```

# PREPARE THE FEATURE & TARGET VARIABLE 

```python
# Convert TotalCharges to numeric values
# Any invalid values will become missing values (NaN)
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

# Store customer IDs separately for later reference
customer_ids = df["customerID"].copy()

# Store the target variable
y = df["Churn"].copy()

# Create the feature dataset
# Remove customerID and Churn from the input features
X = df.drop(
    columns=["customerID", "Churn"]
).copy()

# Display the feature dataset
print("Features shape:", X.shape)

# Display the target dataset
print("Target shape:", y.shape)

# Display target categories
print("\nTarget values:")
print(y.value_counts(dropna=False))
```

```python
# Check whether the target contains missing values
print("Missing target values:", y.isnull().sum())

# Check the unique target categories
print("Unique target values:", y.unique())

# Stop if the target contains missing or unexpected values
if y.isnull().any():
    raise ValueError("The target column contains missing values.")

if not set(y.unique()).issubset({"Yes", "No"}):
    raise ValueError("Unexpected values found in the Churn column.")
```

# SPLIT THE DATASET INTO TRAINING & TEST SETS 

```python
# Split the features and target into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,                  # Input features
    y,                  # Target variable
    test_size=0.20,     # Keep 20% for testing
    random_state=42,    # Make the split reproducible
    stratify=y          # Preserve target class proportions
)

# Display the resulting dataset sizes
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

# IDENTIFY NUMERICAL & CATEGORICAL COLUMNS 

```python
# Identify numerical columns from the training dataset
numerical_features = X_train.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

# Identify categorical columns from the training dataset
categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# Display numerical columns
print("Numerical features:")
print(numerical_features)

# Display categorical columns
print("\nCategorical features:")
print(categorical_features)

# Display the total number of each type
print("\nNumber of numerical features:", len(numerical_features))
print("Number of categorical features:", len(categorical_features))
```

```python
# Combine the two feature lists
identified_features = numerical_features + categorical_features

# Find any columns that were not identified
unidentified_features = [
    col for col in X_train.columns
    if col not in identified_features
]

print("Unidentified columns:", unidentified_features)

# Stop if any columns were not assigned a type
if unidentified_features:
    raise ValueError(
        "Some columns were not identified. Inspect their data types."
    )
```

# BUILD NUMERICAL PREPROCESSING PIPELINE 

```python
# Create the numerical preprocessing pipeline
numerical_pipeline = Pipeline([

    # Fill missing numerical values using the median
    ("imputer", SimpleImputer(strategy="median")),

    # Standardize numerical features
    ("scaler", StandardScaler())

])

# Display the pipeline
numerical_pipeline
```

# BUILD CATEGORICAL PREPROCESSING PIPELINE 

```python
# Create the categorical preprocessing pipeline
categorical_pipeline = Pipeline([

    # Fill missing categorical values using the most frequent category
    ("imputer", SimpleImputer(strategy="most_frequent")),

    # Convert categories into numerical columns
    ("encoder", OneHotEncoder(handle_unknown="ignore"))

])

# Display the pipeline
categorical_pipeline
```

# COMBINING BOTH PIPELINES USING COLUMN TRANSFORMER 

```python
# Combine numerical and categorical preprocessing
preprocessor = ColumnTransformer(
    transformers=[

        # Apply numerical preprocessing to numerical columns
        ("num", numerical_pipeline, numerical_features),

        # Apply categorical preprocessing to categorical columns
        ("cat", categorical_pipeline, categorical_features)

    ],

    # Ignore any columns not explicitly included
    remainder="drop"
)

# Display the complete preprocessing structure
preprocessor
```

# FIT THE PROCESSOR ON TRAINING DATA 

```python
# Learn preprocessing rules from the training data
# and transform the training data
X_train_prepared = preprocessor.fit_transform(X_train)

# Apply the already learned preprocessing rules to the test data
# Do not use fit_transform on the test data
X_test_prepared = preprocessor.transform(X_test)

# Display the shapes after preprocessing
print("Prepared training data shape:", X_train_prepared.shape)

print("Prepared testing data shape:", X_test_prepared.shape)
```

# IMPORT MACHINE LEARNING ALGORITHMS 

```python
# Import Logistic Regression
from sklearn.linear_model import LogisticRegression

# Import Decision Tree
from sklearn.tree import DecisionTreeClassifier

# Import Random Forest
from sklearn.ensemble import RandomForestClassifier

# Import Extra Trees
from sklearn.ensemble import ExtraTreesClassifier

# Import Pipeline
from sklearn.pipeline import Pipeline

# Import clone so each model gets its own independent preprocessor
from sklearn.base import clone

# Import cross-validation tools
from sklearn.model_selection import StratifiedKFold, cross_validate

# Import metric functions
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# Import make_scorer to define churn as the positive class
from sklearn.metrics import make_scorer

# Define the positive class as customers who churn
precision_scorer = make_scorer(
    precision_score,
    pos_label="Yes",
    zero_division=0
)

recall_scorer = make_scorer(
    recall_score,
    pos_label="Yes",
    zero_division=0
)

f1_scorer = make_scorer(
    f1_score,
    pos_label="Yes",
    zero_division=0
)
```

# CREATE A DICTIONARY OF MODULE  

```python
# Create a dictionary containing the machine learning algorithms
models = {

    # Logistic Regression
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=42
    ),

    # Decision Tree
    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    # Random Forest
    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ),

    # Extra Trees
    "Extra Trees": ExtraTreesClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    )
}

print("Models created successfully!")
```

# BUILD A COMPLETE PIPELINE FOR EACH MODULE 

```python
# Create an empty dictionary to store complete model pipelines
model_pipelines = {}

# Loop through every model in the dictionary
for model_name, model in models.items():

    # Create a pipeline with preprocessing and model training
    pipeline = Pipeline([
        ("preprocessing", clone(preprocessor)),
        ("model", model)
    ])

    # Store the pipeline using its model name
    model_pipelines[model_name] = pipeline

print("Complete model pipelines created!")
```

# TRAIN THE MODELS 

```python
# Train every model using the training data
for model_name, pipeline in model_pipelines.items():

    # Fit the complete pipeline
    pipeline.fit(X_train, y_train)

    # Display the training status
    print(model_name, "trained successfully!")
```

# EVALUATE INITIAL MODEL PERFORMANCE 

```python
# Create an empty list to store training results
training_results = []

# Evaluate each trained model on the training data
for model_name, pipeline in model_pipelines.items():

    # Predict churn labels
    predictions = pipeline.predict(X_train)

    # Get the probability of churn ("Yes")
    yes_index = list(pipeline.classes_).index("Yes")
    probabilities = pipeline.predict_proba(X_train)[:, yes_index]

    # Calculate performance metrics
    training_results.append({
        "Model": model_name,
        "Accuracy": accuracy_score(y_train, predictions),
        "Precision": precision_score(
            y_train, predictions, pos_label="Yes", zero_division=0
        ),
        "Recall": recall_score(
            y_train, predictions, pos_label="Yes", zero_division=0
        ),
        "F1 Score": f1_score(
            y_train, predictions, pos_label="Yes", zero_division=0
        ),
        "ROC AUC": roc_auc_score(
            (y_train == "Yes").astype(int), probabilities
        )
    })

# Convert the results into a DataFrame
training_results_df = pd.DataFrame(training_results)

# Display the results
training_results_df.round(3)
```

# * MODEL VALIDATION & COMPARISON  

# CROSS VALIDATION  

```python
# Create stratified 5-fold cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Define the metrics to calculate during cross-validation
scoring = {
    "accuracy": "accuracy",
    "precision": precision_scorer,
    "recall": recall_scorer,
    "f1": f1_scorer,
    "roc_auc": "roc_auc"
}
```

# PERFORM CROSS VALIDATION FOR ALL MODELS 

```python
# Store cross-validation results
cv_results = []

# Evaluate each model pipeline
for model_name, pipeline in model_pipelines.items():

    print("Evaluating:", model_name)

    # Run cross-validation
    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1,
        return_train_score=True
    )

    # Store average validation scores
    cv_results.append({
        "Model": model_name,
        "Accuracy Mean": scores["test_accuracy"].mean(),
        "Precision Mean": scores["test_precision"].mean(),
        "Recall Mean": scores["test_recall"].mean(),
        "F1 Mean": scores["test_f1"].mean(),
        "ROC AUC Mean": scores["test_roc_auc"].mean(),

        # Standard deviation shows variation across folds
        "ROC AUC Std": scores["test_roc_auc"].std(),
        "Recall Std": scores["test_recall"].std()
    })

# Convert the results to a DataFrame
cv_results_df = pd.DataFrame(cv_results)

# Sort models by mean ROC-AUC
cv_results_df = cv_results_df.sort_values(
    by="ROC AUC Mean",
    ascending=False
)

# Display the results
cv_results_df.round(3)
```

# CHOOSE A CHURN PROBABILITY THRESHOLD 

```python
# Import cross_val_predict
from sklearn.model_selection import cross_val_predict

# Generate out-of-fold probabilities using training data only
oof_probabilities = cross_val_predict(
    best_rf_model,
    X_train,
    y_train,
    cv=cv,
    method="predict_proba",
    n_jobs=-1
)

# Find the position of the Yes class
yes_index = list(best_rf_model.classes_).index("Yes")

# Extract churn probabilities
oof_churn_probabilities = oof_probabilities[:, yes_index]

# Convert training labels to binary values
y_train_binary = (y_train == "Yes").astype(int)

# Calculate precision, recall and candidate thresholds
oof_precision, oof_recall, oof_thresholds = (
    precision_recall_curve(
        y_train_binary,
        oof_churn_probabilities
    )
)

# Calculate F1 score for each available threshold
oof_f1_scores = (
    2 * oof_precision[:-1] * oof_recall[:-1]
    / (oof_precision[:-1] + oof_recall[:-1] + 1e-10)
)

# Find the threshold with the highest F1 score
best_threshold_index = np.argmax(oof_f1_scores)

selected_threshold = oof_thresholds[best_threshold_index]

print("Selected threshold:", selected_threshold)
print("Training out-of-fold F1:", oof_f1_scores[best_threshold_index])
print("Training out-of-fold precision:", oof_precision[best_threshold_index])
print("Training out-of-fold recall:", oof_recall[best_threshold_index])
```

# CREATE USABLE PREDICTED FUNCTION 

```python
def predict_customer_churn(new_customer_data):

    # Check that the input is a pandas DataFrame
    if not isinstance(new_customer_data, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")

    # Retrieve the expected feature columns
    expected_columns = loaded_config["feature_columns"]

    # Check for missing columns
    missing_columns = [
        col for col in expected_columns
        if col not in new_customer_data.columns
    ]

    # Check for unexpected columns
    extra_columns = [
        col for col in new_customer_data.columns
        if col not in expected_columns
    ]

    # Stop if the input schema is incorrect
    if missing_columns or extra_columns:
        raise ValueError(
            f"Missing columns: {missing_columns}; "
            f"Unexpected columns: {extra_columns}"
        )

    # Arrange columns in the exact expected order
    new_customer_data = new_customer_data[expected_columns].copy()

    # Predict probabilities using the saved model
    probabilities = loaded_model.predict_proba(
        new_customer_data
    )

    # Find the probability column for Yes
    yes_index = list(loaded_model.classes_).index("Yes")

    # Extract churn probabilities
    churn_probabilities = probabilities[:, yes_index]

    # Apply the saved business threshold
    predictions = np.where(
        churn_probabilities >= loaded_config["prediction_threshold"],
        "Yes",
        "No"
    )

    # Create a results DataFrame
    results = pd.DataFrame({
        "Churn_Probability": churn_probabilities,
        "Predicted_Churn": predictions
    }, index=new_customer_data.index)

    return results
```
